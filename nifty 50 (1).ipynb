```python
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Book 1.csv")
```

```python
data
```

```python
print(data.isnull().sum())
```

```python
print(data.describe())
```

```python
x = np.arange(len(data['DATE']))

plt.figure(figsize=(8,4))

plt.bar(x - 0.2, data[' DAY HIGH'], width=0.4, label='Day High')
plt.bar(x + 0.2, data['DAY LOW'], width=0.4, label='Day Low')

plt.xticks(x, data['DATE'], rotation=90)

plt.xlabel("Date")
plt.ylabel("Price")
plt.title("Date-wise High & Low Bar Chart")

plt.legend()
plt.show()
```

# DATE - WISE HIGH AND LOW LINE CHART

```python
plt.figure(figsize=(12,6))
plt.plot(data[' DAY HIGH'],label='Day High')
plt.plot(data['DAY LOW'],label='Day Low')

plt.xticks(x, data['DATE'], rotation=45)

plt.xlabel("Date")
plt.ylabel("Price")
plt.title("Date-wise High & Low line Chart")

plt.legend()
plt.show()
```

# DATE - WISE (DAY-HIGH)

```python
plt.figure(figsize=(12,6))
plt.bar(data['DATE'], data[' DAY HIGH'])
plt.xlabel("Date")
plt.ylabel("Day High")
plt.title("Date Wise Day High")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# DATE WISE (DAY-LOW)

```python
plt.figure(figsize=(8,4))
plt.bar(data['DATE'], data['DAY LOW'])
plt.xlabel("Date")
plt.ylabel("Day Low")
plt.title("Date Wise Day LOW")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
